### Data Cleaning

#### 1. Data NaN

```python
# Cek jumlah missing value per kolom
df.isnull().sum()
```

```python
df['cancellation']= df['cancellation'].fillna('Not Cancelled')
```

```python
# Cari indeks dimana order_time adalah NaN
nan_indices = df[df['order_time'].isna()].index.tolist()

# Tampilkan 5 data sebelum dan sesudah NaN
for idx in nan_indices[:1]:
    print(f"\n{'='*80}")
    print(f"NaN ditemukan di indeks: {idx}")
    print(f"{'='*80}")
    
    # Tentukan range untuk diambil (5 sebelum dan 5 sesudah)
    start = max(0, idx - 5)
    end= min(len(df), idx + 6)
    
    # Ambil data di range tersebut
    context_df = df.iloc[start:end].copy()
    
    # Tandai baris yang NaN
    context_df['is_nan'] = context_df['order_time'].isna()
    
    # Tampilkan kolom yang relevan
    display_cols = ['order_id', 'order_time']
    print(context_df[display_cols + ['is_nan']].to_string())
```

```python
df_filled= df.copy()
df_filled['order_time']= df_filled['order_time'].fillna(method='ffill')
```

#### 2. Data Duplikat

```python
# cek duplikat
df_filled.duplicated().sum()
```

```python
df_filled= df_filled.drop_duplicates()
```

#### 3. cek tipe data

```python
df_filled['order_time'] = pd.to_datetime(df_filled['order_time'])
```

```python
# cek kolom yang cocok dibuat category, yg jml nilai unique 5% total data

print(df_filled['payment_method'].unique())

print(f"Jumlah payment_method: {df_filled['payment_method'].nunique()}")
print(f"Jumlah shipping_option: {df_filled['shipping_option'].nunique()}")
print(f"Jumlah order_status: {df_filled['order_status'].nunique()}")
print(f"Jumlah city: {df_filled['city'].nunique()}")
print(f"Jumlah province: {df_filled['province'].nunique()}")
print(f"Jumlah cancellation: {df_filled['cancellation'].nunique()}")
print(f"Jumlah source_file: {df_filled['source_file'].nunique()}")
print(f"Jumlah product_categories: {df_filled['product_categories'].nunique()}")
```

```python
# ubah jadi category
for col in ['order_status', 'shipping_option', 'payment_method', 'city', 'province', 'cancellation', 'source_file']:
    df_filled[col]= df_filled[col].astype('category')
```

#### cek data

```python
# 2. Cek apakah total_payment >= 0 (tidak boleh negatif)
df_filled[df_filled['total_payment'] < 0]
```

```python
# 3. Cek apakah total_qty > 0
df_filled[df_filled['total_qty'] <= 0]
```

### Data Preprocessing

```python
df_filled['year']= df_filled['order_time'].dt.year
```

```python
df_filled['month'] = df_filled['order_time'].dt.month
```

```python
df_filled['day_of_week'] = df_filled['order_time'].dt.day_name().astype('category')
```

```python
df_filled['profit']= df_filled['total_payment'] - df_filled['shipping_cost']
```

```python
df_filled['avg_item_price']= df_filled['total_payment']/ df_filled['total_qty']
```

```python
df_filled['return_rate']= (df_filled['total_returned_qty']/ df_filled['total_qty'])*100
```

```python
def segment_order(qty):
    if qty == 1:
        return 'Single'
    elif qty <= 3:
        return 'Small'
    elif qty <= 6:
        return 'Medium'
    else:
        return 'Large'
    
df_filled['order_segment']= df_filled['total_qty'].apply(segment_order).astype('category')
```
