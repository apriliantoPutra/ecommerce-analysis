### Data Cleaning

#### 1. Data NaN

```python
# Cek jumlah missing value per kolom
df.isnull().sum()
```

```python
# Isi data kosong pada kolom 'cancellation' dengan 'Not Cancelled'
df['cancellation']= df['cancellation'].fillna('Not Cancelled')
```

```python
# cek data kosong pada kolom 'order_time'

nan_indices = df[df['order_time'].isna()].index.tolist()

# Tampilkan 5 data sebelum dan sesudah NaN
for idx in nan_indices[:1]:
    print(f"\n{'='*80}")
    print(f"NaN ditemukan di indeks: {idx}")
    print(f"{'='*80}")
    
    # Tentukan range untuk diambil (5 sebelum dan 5 sesudah)
    start = max(0, idx - 5)
    end= min(len(df), idx + 6)
    
    # Ambil data di range tersebut
    context_df = df.iloc[start:end].copy()
    
    # Tandai baris yang NaN
    context_df['is_nan'] = context_df['order_time'].isna()
    
    # Tampilkan kolom yang relevan
    display_cols = ['order_id', 'order_time', 'source_file']
    print(context_df[display_cols + ['is_nan']].to_string())
```

```python
# Isi kolom 'order_time' secara merata

def extract_month_year(filename):
    # Pola: NamaBulanTahunSalesTahun.xlsx
    # Contoh: DecemberSales2023.xlsx, JulySales2025.xlsx
    
    # Ekstrak bulan
    months = {
        'January': 1, 'February': 2, 'March': 3, 'April': 4,
        'May': 5, 'June': 6, 'July': 7, 'August': 8,
        'September': 9, 'October': 10, 'November': 11, 'December': 12
    }
    
    for month_name, month_num in months.items():
        if month_name in filename:
            # Ekstrak tahun (4 digit angka setelah bulan)
            year_match = re.search(rf'{month_name}Sales(\d{{4}})\.xlsx', filename)
            if year_match:
                year = int(year_match.group(1))
                return month_num, year
    return None, None

# Fungsi untuk membuat tanggal dan waktu yang merata
def generate_balanced_dates(start_date, end_date, n):
    """Menghasilkan n tanggal yang merata antara start_date dan end_date"""
    if n <= 1:
        return [start_date + timedelta(hours=12)]  # Tengah hari jika hanya 1 data
    
    # Buat waktu yang merata antara start_date dan end_date
    total_seconds = (end_date - start_date).total_seconds()
    interval = total_seconds / n
    
    dates = []
    for i in range(n):
        # Mulai dari start_date + (interval * i) + sedikit variasi
        random_offset = np.random.randint(-3600, 3601)  # Variasi ±1 jam
        new_date = start_date + timedelta(seconds=interval * i + random_offset)
        # Pastikan tidak melewati end_date
        if new_date > end_date:
            new_date = end_date - timedelta(hours=1)
        if new_date < start_date:
            new_date = start_date + timedelta(hours=1)
        dates.append(new_date)
    
    return dates

# 1. Identifikasi semua source_file yang unik
unique_sources = df['source_file'].unique()

# 2. Untuk setiap source_file, proses
for source in unique_sources:
    # Filter data untuk source_file ini
    mask = df['source_file'] == source
    source_data = df[mask]
    
    # Cek apakah ada order_time yang NaN
    nan_mask = source_data['order_time'].isna()
    if nan_mask.any():
        # Ekstrak bulan dan tahun dari nama file
        month, year = extract_month_year(source)
        
        if month and year:
            # Tentukan jumlah data yang kosong
            n_nan = nan_mask.sum()
            
            # Buat range tanggal untuk bulan tersebut
            # Tentukan jumlah hari dalam bulan
            if month == 2:  # Februari
                if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
                    days_in_month = 29
                else:
                    days_in_month = 28
            elif month in [4, 6, 9, 11]:
                days_in_month = 30
            else:
                days_in_month = 31
            
            # Buat tanggal mulai dan akhir (1-31 atau sesuai jumlah hari)
            start_date = datetime(year, month, 1, 0, 0, 0)
            end_date = datetime(year, month, days_in_month, 23, 59, 59)
            
            # Generate tanggal yang merata
            new_dates = generate_balanced_dates(start_date, end_date, n_nan)
            
            # Urutkan dates agar lebih rapi
            new_dates.sort()
            
            # Isi nilai NaN dengan tanggal yang telah dibuat
            nan_indices = df[mask & nan_mask].index
            for idx, new_date in zip(nan_indices, new_dates):
                df.loc[idx, 'order_time'] = new_date
            
            print(f"✅ {source}: Mengisi {n_nan} data kosong dengan tanggal merata dari {start_date.date()} sampai {end_date.date()}")
        else:
            print(f"⚠️ Tidak bisa mengekstrak bulan/tahun dari: {source}")
```

### Data Preprocessing

```python
df['month_period'] = df['order_time'].dt.to_period('M')
```

```python
df['day_of_week']= df['order_time'].dt.day_name()
```

```python
df['profit']= df['total_payment'] - df['shipping_cost']
```

```python
def segment_order(qty):
    if qty == 1:
        return 'Single'
    elif qty <= 3:
        return 'Small'
    elif qty <= 6:
        return 'Medium'
    else:
        return 'Large'
    
df['order_segment']= df['total_qty'].apply(segment_order)
```

```python
def segment_dis(dis):
    if dis == 0:
        return 'WithOut Discount'
    elif dis >= 1 and dis <= 10000:
        return '1-10k'
    elif dis > 10000 and dis <= 50000:
        return '10k-50k'
    elif dis > 50000 and dis <= 100000:
        return '50k-100k'
    elif dis > 100000 and dis <= 500000:
        return '100k-500k'
    elif dis > 500000:
        return '>500k'
    else:
        return 'other'
    
df['discount_segment']= df['total_discount'].apply(segment_dis)
```
